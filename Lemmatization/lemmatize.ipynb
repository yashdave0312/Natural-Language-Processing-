WordNet Lemmatizer 

```python
from nltk.stem import WordNetLemmatizer

lematizer = WordNetLemmatizer()

words = ["going","gone","goes","running","runs","ran","runner","walked","walking","walks","history"]

for word in words:
    print(f"{word} --> {lematizer.lemmatize(word)}")
```

```python
#  Here we can see that the lemmatizer is not able to lemmatize the words properly. It is not able to lemmatize the words like "going", "gone", "goes", "running", "runs", "ran", "runner", "walked", "walking", "walks". It is only able to lemmatize the word "history" correctly. This is because the lemmatizer needs the part of speech (POS) of the word to lemmatize it correctly. By default, it assumes that the word is a noun.

# Different pos tags are:
# a for adjective
# v for verb
# n for noun
# r for adverb

for word in words:
    print(f"{word} --> {lematizer.lemmatize(word,pos = "v")}")
```

Hence we can conclude that Lemmatization is much better than Stemming
