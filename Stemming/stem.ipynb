Stemming

1) Porter Stemmer 

```python
from nltk.stem import PorterStemmer

words = ["going","gone","goes","running","runs","ran","runner","walked","walking","walks","history"]

port_stem = PorterStemmer()

for word in words :
    print(f"{word} ---> {port_stem.stem(word)}")    
```

2) RegExpStemmer Class

```python
from nltk.stem import RegexpStemmer

re_stem =RegexpStemmer("ing$|es$|s$|ed$", min =4)

for word in words :
    print(f"{word} ---> {re_stem.stem(word)}")  
```

3) SnowBall Stemmer

```python
from nltk.stem import SnowballStemmer

snow_stem = SnowballStemmer("english")

for word in words :
    print(f"{word} ---> {snow_stem.stem(word)}")  
```
